## Chest X-Ray Pneumonia Classification Project

### 1. Dataset Selection & Metadata
* **Dataset:** Chest X-Ray Images (Pneumonia) (`paultimothymooney/chest-xray-pneumonia`)
* **License:** CC BY 4.0
* **Task Type:** Binary Image Classification (`NORMAL` vs. `PNEUMONIA`)
* **Domain Importance:** Automated detection of pneumonia from chest radiographs assists clinicians in rapid screening, reducing diagnostic delays and minimizing false negatives.

```python
import os
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

# Download dataset
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
base_dir = os.path.join(path, "chest_xray")
print("Dataset path:", base_dir)
```

### 2. Exploratory Data Analysis (EDA) - Class Distribution

```python
splits = ['train', 'val', 'test']
classes = ['NORMAL', 'PNEUMONIA']

data = []
for split in splits:
    for cls in classes:
        cls_dir = os.path.join(base_dir, split, cls)
        count = len([f for f in os.listdir(cls_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
        data.append({'Split': split, 'Class': cls, 'Count': count})

df_counts = pd.DataFrame(data)
print(df_counts)

plt.figure(figsize=(8, 5))
sns.barplot(data=df_counts, x='Split', y='Count', hue='Class')
plt.title('Class Distribution Across Splits')
plt.ylabel('Number of Images')
plt.show()
```

### 3. Visual Inspection / Sample Grid

```python
fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for row_idx, cls in enumerate(classes):
    cls_dir = os.path.join(base_dir, 'train', cls)
    sample_files = [f for f in os.listdir(cls_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))][:4]
    for col_idx, fname in enumerate(sample_files):
        img_path = os.path.join(cls_dir, fname)
        img = Image.open(img_path).convert('L')
        axes[row_idx, col_idx].imshow(img, cmap='gray')
        axes[row_idx, col_idx].set_title(f"{cls}")
        axes[row_idx, col_idx].axis('off')

plt.tight_layout()
plt.show()
```
